```python
# 1. Check Positive zero or negative
a = int(input("Enter a number: "))
if a>0:
  print("Positive Number")
elif a==0:
  print("Zero")
else:
  print("Negative")
```

```python
# 2. Greatest number
a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))
c = int(input("Enter 3rd number: "))

if a > b and a > c:
    print(a, "is greater")
elif b > a and b > c:
    print(b, "is greater")
else:
    print(c, "is greater")
```

```python
# 2. Count numbers
count = 0
n = int(input("Enter a number: "))
for i in range(1,n+1):
  if i%3==0:
    count += 1
print(count)
```

```python
# 3. Factorial
fact = 1
n = int(input("Enter a number: "))
for i in range(1,n+1):
  fact *= i
print(fact)
```

# Loops

```python
# 1. Sum of Even Numbers
sum = 0
for i in range(1, 100+1):
  if i%2==0:
    sum += i
print(sum)
```

```python
# 2. Count Digits
n = int(input("Enter a number: "))
count = 0
while n>0:
  n = n//10
  count += 1
print(count)
```

```python
# 3. Reverse a number
n = int(input("Enter a number: "))

temp = n
rev = 0

while temp > 0:
    rem = temp % 10
    rev = rev * 10 + rem
    temp = temp // 10

print(rev)
```

```python
# 4. Mini challenge
n = int(input("Enter a number: "))

temp = n
count = 0
sum_digits = 0
rev = 0

while temp > 0:
    digit = temp % 10

    count += 1
    sum_digits += digit
    rev = rev * 10 + digit

    temp = temp // 10

print("Number of digits:", count)
print("Sum of digits:", sum_digits)
print("Reverse:", rev)
```

```python
# 5. Count even and odd digits
n = int(input("Enter a number: "))
even = 0
odd = 0
while n>0:
  rem = n%10
  if rem%2==0:
    even += 1
  else:
    odd += 1
  n = n//10
print("Even number:", even)
print("Odd Numbers:", odd)
```
