# Loops

```python
# 1. Sum of Even Numbers
sum = 0
for i in range(1, 100+1):
  if i%2==0:
    sum += i
print(sum)
```

```python
# 2. Count Digits
n = int(input("Enter a number: "))
count = 0
while n>0:
  n = n//10
  count += 1
print(count)
```

```python
# 3. Reverse a number
n = int(input("Enter a number: "))

temp = n
rev = 0

while temp > 0:
    rem = temp % 10
    rev = rev * 10 + rem
    temp = temp // 10

print(rev)
```

```python
# 4. Mini challenge
n = int(input("Enter a number: "))

temp = n
count = 0
sum_digits = 0
rev = 0

while temp > 0:
    digit = temp % 10

    count += 1
    sum_digits += digit
    rev = rev * 10 + digit

    temp = temp // 10

print("Number of digits:", count)
print("Sum of digits:", sum_digits)
print("Reverse:", rev)
```

```python
# 5. Count even and odd digits
n = int(input("Enter a number: "))
even = 0
odd = 0
while n>0:
  rem = n%10
  if rem%2==0:
    even += 1
  else:
    odd += 1
  n = n//10
print("Even number:", even)
print("Odd Numbers:", odd)
```

```python
# 6. Find the largest digit in a number.
n = int(input("Enter a number: "))
largest = 0
while n>0:
  temp = n%10
  if temp>largest:
    largest = temp
  n = n//10
print(largest)
```

```python
# 7. Find the smallest digit in a number.
n = int(input("Enter a number: "))
print(min(str(n)))
small = 9
while n>0:
  dig = n%10
  if dig<small:
    small = dig
  n = n//10
print(small)
```

```python
# 8. Seconf largest
n = int(input("Enter a number: "))

largest = 0
secLar = 0

while n > 0:
    temp = n % 10

    if temp > largest:
        secLar = largest
        largest = temp

    elif temp > secLar:
        secLar = temp

    n = n // 10

print("Largest:", largest)
print("Second largest:", secLar)
```
