```python
# 1. Check Positive zero or negative
a = int(input("Enter a number: "))
if a>0:
  print("Positive Number")
elif a==0:
  print("Zero")
else:
  print("Negative")
```

```python
# 2. Greatest number
a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))
c = int(input("Enter 3rd number: "))

if a > b and a > c:
    print(a, "is greater")
elif b > a and b > c:
    print(b, "is greater")
else:
    print(c, "is greater")
```

```python
# 3. Multiple
a = int(input("Enter a number: "))
for i in range(1, 11):
  print(a, "X", i, "=", a*i)
```

```python
# 4. Divisible by 2 from 1 to 100
for i in range(1,101):
  if i%2==0:
    print(i)
```

```python
# 1. Sum 1 to 100
sum = 0
for i in range(1,101):
  sum = sum + i
print(sum)
```

```python
# 2. Count numbers
count = 0
n = int(input("Enter a number: "))
for i in range(1,n+1):
  if i%3==0:
    count += 1
print(count)
```

```python
# 3. Factorial
fact = 1
n = int(input("Enter a number: "))
for i in range(1,n+1):
  fact *= i
print(fact)
```
