# Objects and Classes


```python
class Human:
  def __init__(self, age, name):
    self._age = age
    self._name = name
  def __str__(self):
    return "Hi"
h = Human(age=4, name="Bob the builder")
print(h)
```

```python
class Human:
  def __init__(self, age, name):
    self._age = age
    self._name = name

  def __str__(self):
    return "A human with name" + self._name + "." + "Their age is" + str(self._age) +"."

h = Human(age=20, name="Ved")
h
```

```python
class Human:
    def __init__(self, age, name):
        self._age = age
        self._name = name

    def __str__(self):
        return f"A human with name {self._name}. Their age is {self._age}."

    '''def __repr__(self):
        return self.__str__()'''

h = Human(20, "Ved")
h
```

#### SET Operations

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(A | B)
A.union(B) # Combines both sets
print(A & B) # Intersection
print( A - B) # Difference
print(A ^ B) # Symmetric Difference
```
